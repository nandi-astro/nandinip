```python
arr = np.array([
    [2, 3, 5],
    [14, 16, 8],
    [11, 13, 17],
    [7, 10, 13]
])

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
        else:
            return True

def prime_rows(array):
    prime_rows = []
    for row in arr:
        if any(is_prime(x) for x in row):
            prime_rows.append(row)
    return np.array(prime_rows)

prime_star_rows = prime_rows(arr)
print(prime_star_rows)
```

```python
def eightby():
    return np.zeros((8,8), dtype=int)
eightby()
```

```python
arr = eightby()

for i in range(0, 8, 2):
    arr[i, ::2] = 1

print(arr)
```

```python
def checkerboard():
    arr = eightby()
    arr[1::2, ::2] = 1
    arr[::2, 1::2] = 1
    return arr

checkerboard()
```

```python
def reverse_checkerboard():
    arr = eightby()
    arr[::2, ::2] = 1
    arr[1::2, 1::2] = 1
    return arr

reverse_checkerboard()
```

```python
def expansion():
    start_string = input("Type a word or phrase")
    input_spaces = int(input("How many spaces between letters?"))
    space_separator = " " * input_spaces
    return space_separator.join(start_string)

expansion()
```

```python
def secondBrightest(array):
    guidestars = np.array([])
    for row in array:
        sorted_row = np.sort(row)[::1]
        guidestars = np.append(guidestars, sorted_row.item(1))
    return guidestars
```
