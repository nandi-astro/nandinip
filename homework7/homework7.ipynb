```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    y = A*np.sin(w*x)
    return y
```

```python
x = np.linspace(0, 2*np.pi, 1000)
```

```python
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]
```

```python
fig, axes = plt.subplots(1, 3, figsize=(25,5))

#left subplot
for amp in amplitude:
    axes[0].plot(x, make_sine_wave(x, A=amp, w=1), label=f"A{amp}")
axes[0].set_title("Vary Amplitude (freq=1)")
#middle subplot
for freq in frequency:
    axes[1].plot(x, make_sine_wave(x, A=1, w=freq), label=f"F={freq}")
axes[1].set_title("Vary Frequency (amp=1)")
#right subplot
for amp, freq in zip(amplitude, frequency):
    axes[2].plot(x, make_sine_wave(x, A=amp, w=freq), label=f"A={amp}, F={freq}")
axes[2].set_title("Vary Amplitude and Frequency Together")

for a in axes:
    a.legend()

plt.show()
```
